IMPLEMENTACION DE LA MPNEURON

```python
from flask import Flask, request, jsonify, make_response
import numpy as np

app = Flask(__name__)

class MPNeuron:
    def __init__(self):
        self.threshold = None  
    
    def model(self, x):
        z = sum(x)
        return (z >= self.threshold)
    
    def predict(self, X):  
        Y = []
        for x in X:
            result = self.model(x)
            Y.append(result)
        return np.array(Y)

mp_neuron = MPNeuron()
mp_neuron.threshold = 4

# Definimos la ruta "/predict" que acepta peticiones POST y OPTIONS
@app.route("/predict", methods=["POST", "OPTIONS"])
def predict():
    # Si la petición es de tipo OPTIONS (preflight de CORS)
    if request.method == "OPTIONS":
        response = make_response()  # Creamos una respuesta vacía
        # Permitimos acceso desde el frontend en http://127.0.0.1:5500
        response.headers["Access-Control-Allow-Origin"] = "http://127.0.0.1:5500"
        # Permitimos que se envíe la cabecera Content-Type
        response.headers["Access-Control-Allow-Headers"] = "Content-Type"
        # Indicamos que solo se permiten los métodos POST y OPTIONS
        response.headers["Access-Control-Allow-Methods"] = "POST, OPTIONS"
        return response  # Devolvemos la respuesta al navegador

    # Si la petición es POST normal (con datos)
    # Obtenemos del body JSON el campo "inputs"
    data = request.json.get("inputs")
    # Usamos el modelo mp_neuron para hacer predicciones con esos datos
    predictions = mp_neuron.predict(data)
    # Convertimos el resultado a JSON (predictions probablemente es un array de NumPy)
    response = jsonify({"predictions": predictions.tolist()})
    # Permitimos acceso desde el frontend en http://127.0.0.1:5500
    response.headers["Access-Control-Allow-Origin"] = "http://127.0.0.1:5500"
    return response  # Devolvemos la respuesta con las predicciones

# Si ejecutamos este archivo directamente, iniciamos el servidor Flask
if __name__ == "__main__":
    # Levantamos la app en el puerto 5000, con debug activado y sin autoreload
    app.run(debug=True, port=5000, use_reloader=False)
```
